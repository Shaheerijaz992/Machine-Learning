```python
import numpy as np
A=[1,4,5]
B=[3,7,8]
A_matrix = [[1, 2, 3],[4, 5, 6]]
B_matrix = [[7,  8],[9, 10],[11, 12]]
Result=[]
#25.Write a small Python function, using a loop, that computes the dot product of two vectors.
def dot_product():
    for x in range(len(A)):
        product = A[x] * B[x]
        Result.append(product)
    result=sum(Result)
    print(f"Dot Product Using Loop:{result}")
dot_product()
```

```python
#26.Compute the same dot product using NumPy's built in function and confirm both answers match exactly.
dot_np=np.dot(A,B)
print(f"Dot product using numpy:\n{dot_np}")
```

```python
#27.	Write a function using nested loops that multiplies two small matrices together by hand
def matrix_multiplication():
    result = [[0, 0], [0, 0]]
    for i in range(2):
        for j in range(2):
            for k in range(3):
                result[i][j] += A_matrix[i][k] * B_matrix[k][j]
    print("Matrix Multiplication USing Loops:")
    print(result)
matrix_multiplication()
```

```python
#28.	Compute the same result using NumPy and compare both the answer and how much shorter the NumPy code is.
multiply_result=np.matmul(A_matrix,B_matrix)
print(f"Matrix Multiplication USing Numpy:\n{multiply_result}")
```

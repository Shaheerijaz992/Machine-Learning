```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
# Load or create a small dataset containing house size and house price columns.
data = {
    "Size": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500],
    "Price": [150000, 180000, 210000, 260000, 300000, 330000, 360000, 400000]
}
df=pd.DataFrame(data)
x=df[["Size"]]
y=df[["Price"]]
```

```python
# Load or create a small dataset containing house size and house price columns.
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Create a LinearRegression model and fit it using only the training data.
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
# Use the trained model to predict prices for the houses in the testing set.
y_pred=model.predict(x_test)
mae=mean_absolute_error(y_test,y_pred)
```

```python
# Compare the predictions against the true prices using mean absolute error, and plot the model's predicted line against the actual data points.
print("Actual Prices:")
print(y_test.values)
print("Predicted Prices:")
print(y_pred)
print("Mean Absolute Error:", mae)
plt.scatter(x,y,label="Actual Data")
plt.plot(x,model.predict(x),label="Regression Line")
plt.xlabel("House Size")
plt.ylabel("House Price")
plt.title("House Price Prediction")
plt.show
```

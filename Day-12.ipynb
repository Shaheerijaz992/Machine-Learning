```python
import pandas as pd
import seaborn as sb
# Load Titanic dataset
titanic = sb.load_dataset("titanic")
# 1. Count missing values
print("Missing values:")
print(titanic.isnull().sum())
# 2. Fill missing age with median
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
# 3. Drop deck because too many values are missing
titanic = titanic.drop("deck", axis=1)
# 4. Fill missing embarked values with mode
titanic["embarked"] = titanic["embarked"].fillna(
    titanic["embarked"].mode()[0])
# 5. Convert sex into numbers
titanic["sex"] = titanic["sex"].map({
    "male": 0,
    "female": 1
})
# 6. Create family size
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1
# 7. Check missing values again
print("\nMissing values after cleaning:")
print(titanic.isnull().sum())
# 8. Look at the cleaned dataset
print("\nCleaned dataset:")
print(titanic.head())
```

```python
#Split your data into training and testing sets first, before any scaling happens.
from sklearn.model_selection import train_test_split
x=titanic[["pclass","sex","age","family_size"]]
y=titanic["survived"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Fit a StandardScaler using only the training data.
# Use that same fitted scaler to transform both the training data and the testing data.
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
# Retrain KNN on the scaled data and compare its accuracy to the unscaled version from Day 9.
knn_scaled_model=KNeighborsClassifier()
knn_scaled_model.fit(x_train_scaled,y_train)
knn_scaled_pred = knn_scaled_model.predict(x_test_scaled)
accuracy_scaled_knn=accuracy_score(y_test,knn_scaled_pred)
print(f"Scaled KNN Accuracy: {accuracy_scaled_knn * 100:.2f}%")



```

```python
#Retrain a Decision Tree on the same scaled data and confirm for yourself that its accuracy barely
#changes, if at all.
from sklearn.tree import DecisionTreeClassifier
decision_scaled_model = DecisionTreeClassifier()
decision_scaled_model.fit(x_train_scaled,y_train)
decision_scaled_pred = decision_scaled_model.predict(x_test_scaled)
decision_scaled_accuracy = accuracy_score(y_test, decision_scaled_pred)
print(f"Scaled Decision Tree Accuracy: {decision_scaled_accuracy * 100:.2f}%")
```

```python
import pandas as pd
import seaborn as sb
# Load Titanic dataset
titanic = sb.load_dataset("titanic")
# 1. Count missing values
print("Missing values:")
print(titanic.isnull().sum())
# 2. Fill missing age with median
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
# 3. Drop deck because too many values are missing
titanic = titanic.drop("deck", axis=1)
# 4. Fill missing embarked values with mode
titanic["embarked"] = titanic["embarked"].fillna(
    titanic["embarked"].mode()[0])
# 5. Convert sex into numbers
titanic["sex"] = titanic["sex"].map({
    "male": 0,
    "female": 1
})
# 6. Create family size
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1
# 7. Check missing values again
print("\nMissing values after cleaning:")
print(titanic.isnull().sum())
# 8. Look at the cleaned dataset
print("\nCleaned dataset:")
print(titanic.head())
```

```python
#11. Generate the confusion matrix for your best model using scikit learn.
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
x=titanic[["pclass","sex","age","family_size"]]
y=titanic["survived"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
knn_model=KNeighborsClassifier(n_neighbors=7)
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
cm=confusion_matrix(y_test,knn_pred)
print("Confusion Matrix:")
print(cm)
```

```python
#12. Compute its precision, recall, and F1 score.
from sklearn.metrics import precision_score, recall_score, f1_score
ps=precision_score(y_test,knn_pred)
print(f"Precision Score: {ps}")
rs=recall_score(y_test,knn_pred)
print(f"Recall Score: {rs}")
f1=f1_score(y_test,knn_pred)
print(f"F1 Score: {f1}")
```

```python
#13. Plot the ROC curve for the model and compute its AUC score.
from sklearn.metrics import roc_curve, roc_auc_score
knn_probability = knn_model.predict_proba(x_test)[:, 1]
fpr, tpr, thresholds = roc_curve(y_test, knn_probability)
auc_score = roc_auc_score(y_test, knn_probability)
print(f"AUC Score: {auc_score}")
```

```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
# LOAD TITANIC DATASET
titanic = sb.load_dataset("titanic")
# Fill missing age values with median
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
# Drop deck because too many values are missing
titanic = titanic.drop("deck", axis=1)
# Fill missing embarked values with mode
titanic["embarked"] = titanic["embarked"].fillna(
    titanic["embarked"].mode()[0]
)
# Convert gender into numbers
titanic["sex_encoded"] = titanic["sex"].map({
    "male": 0,
    "female": 1
})
# Create family size
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1
# Check the cleaned data
print(titanic.head())
print("\nMissing values:")
print(titanic.isnull().sum())
```

```python
# 1. HISTOGRAM
# Passenger Age Distribution
sb.histplot(data=titanic, x="age", bins=20)
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# 2. Survival Rate by Passenger Class
sb.barplot(data=titanic, x="pclass", y="survived")
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# 3.Survival Rate by Gender
sb.barplot(data=titanic, x="sex", y="survived")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sb.scatterplot(data=titanic, x="age", y="fare")
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

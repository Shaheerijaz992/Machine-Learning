```python

import numpy as np
import pandas as pd
import seaborn as sb
titanic=sb.load_dataset("titanic")
# 21. Manual calculations using NumPy
fare=titanic["fare"]
mean=(np.mean(fare))
print(f"Mean: {mean}")
median_np=(np.median(fare))
print(f"Median: {median_np}")
std=(np.std(fare))
print(f"Standard Deviation: {std}")
variation_np=(np.var(fare))
print(f"Variation: {variation_np}")
```

```python
#22.	Compare your manual results against DataFrame.describe() and confirm they match.
print("Pandas Describe")
print(fare.describe())
print("Pandas Variance")
print(fare.var())
```

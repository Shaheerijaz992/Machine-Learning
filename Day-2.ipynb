```python
import pandas as pd
import seaborn as sb
# 6. Load the Titanic dataset using seaborn's built in dataset loader.
titanic=sb.load_dataset("titanic")
```

```python
#7.	Print head, info, and describe, and read every line of the output rather than skimming it.
print(titanic.head())
titanic.info()
print(titanic.describe())
```

```python
#8.	Filter the dataset to passengers older than thirty,
# then separately to only female passengers, and compare survival counts between the two groups.
older_than_30=titanic[titanic["age"]>30]
print(len(older_than_30))
print(older_than_30["survived"].value_counts())
female=titanic[titanic["sex"]=="female"]
print(len(female))
print(female["survived"].value_counts())
```

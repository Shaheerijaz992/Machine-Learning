```python
import pandas as pd
import seaborn as sb
# Load Titanic dataset
titanic = sb.load_dataset("titanic")
# 1. Count missing values
print("Missing values:")
print(titanic.isnull().sum())
# 2. Fill missing age with median
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
# 3. Drop deck because too many values are missing
titanic = titanic.drop("deck", axis=1)
# 4. Fill missing embarked values with mode
titanic["embarked"] = titanic["embarked"].fillna(
    titanic["embarked"].mode()[0])
# 5. Convert sex into numbers
titanic["sex"] = titanic["sex"].map({
    "male": 0,
    "female": 1
})
# 6. Create family size
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1
# 7. Check missing values again
print("\nMissing values after cleaning:")
print(titanic.isnull().sum())
# 8. Look at the cleaned dataset
print("\nCleaned dataset:")
print(titanic.head())
```

```python
#Train a KNeighborsClassifier, choosing a starting number of neighbors.
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
x=titanic[["pclass","sex","age","family_size"]]
y=titanic["survived"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
knn_model=KNeighborsClassifier()
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
knn_accuracy = accuracy_score(y_test, knn_pred)
accuracy_knn_percent=knn_accuracy*100
print(f"KNN Accuracy Percentage:{accuracy_knn_percent}")
```

```python
#Train a DecisionTreeClassifier on the same training data.
from sklearn.tree import DecisionTreeClassifier
decision_model=DecisionTreeClassifier()
decision_model.fit(x_train,y_train)
decision_pred=decision_model.predict(x_test)
decision_accuracy = accuracy_score(y_test, decision_pred)
accuracy_decision_percent=decision_accuracy*100
print(f"Decision Tree Accuracy Percentage:{accuracy_decision_percent}")
```

```python
# Logistic Regression
logistic_model = LogisticRegression()
logistic_model.fit(x_train, y_train)
logistic_pred = logistic_model.predict(x_test)
logistic_accuracy = accuracy_score(y_test, logistic_pred)
accuracy_logistic_percent=logistic_accuracy*100
print(f"Logistic Regression Accuracy Percentage:{accuracy_decision_percent}")
```

```python
#Try two or three different values for the number of neighbors in KNN and note how the accuracy
#changes each time.
knn_model=KNeighborsClassifier(n_neighbors=3)
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
knn_accuracy = accuracy_score(y_test, knn_pred)
accuracy_knn_percent=knn_accuracy*100
print(f"KNN Accuracy Percentage with k=3:{accuracy_knn_percent}")
knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
knn_accuracy = accuracy_score(y_test, knn_pred)
accuracy_knn_percent=knn_accuracy*100
print(f"KNN Accuracy Percentage with k=5:{accuracy_knn_percent}")
knn_model=KNeighborsClassifier(n_neighbors=7)
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
knn_accuracy = accuracy_score(y_test, knn_pred)
accuracy_knn_percent=knn_accuracy*100
print(f"KNN Accuracy Percentage with k=7 :{accuracy_knn_percent}")
```

```python
#11. Load your chosen dataset and explore it using head, info, and describe.
import pandas as pd
from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()
cancer=pd.DataFrame(data.data,columns=data.feature_names)
cancer["target"]=data.target
print("Missing Values:")
print(cancer.isnull().sum())
print("First Five Rows:")
print(cancer.head())
print("Data Set Information")
print(cancer.info())
print("DataSet Statistics:")
print(cancer.describe())



```

```python
#13. Train at least three different classifiers on it, drawing on the models you practiced this week.
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
x=cancer.drop("target",axis=1)
y=cancer["target"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
#KNeighborsClassifier
knn_model=KNeighborsClassifier()
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
#LogisticRegression
logistic_model=LogisticRegression(max_iter=5000)
logistic_model.fit(x_train,y_train)
logistic_pred=logistic_model.predict(x_test)
#DecisionTreeClassifier
decision_model=DecisionTreeClassifier()
decision_model.fit(x_train,y_train)
decision_pred=decision_model.predict(x_test)
```

```python
#14. Evaluate every model using a confusion matrix, precision, recall, F1 score, and cross validation.
from sklearn.metrics import confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.model_selection import cross_val_score
cm_knn = confusion_matrix(y_test, knn_pred)
print(f"KNN Confusion Matrix: \n{cm_knn}")
precision_knn = precision_score(y_test, knn_pred)
print(f"Precision Score of KNN: {precision_knn}")
recall_knn = recall_score(y_test, knn_pred)
print(f"Recall Score of KNN: {recall_knn}")
f1_knn = f1_score(y_test, knn_pred)
print(f"F1 Score of KNN: {f1_knn}")
knn_cv_scores=cross_val_score(knn_model,x,y,cv=5)
print("KNeighborsClassifier 5-Fold CV Scores:", knn_cv_scores)
print("KNeighborsClassifier CV Accuracy:", knn_cv_scores.mean())
cm_logistic = confusion_matrix(y_test, logistic_pred)
print(f"\nLogistic Regression Confusion Matrix:")
print(cm_logistic)
precision_logistic = precision_score(y_test, logistic_pred)
print(f"Precision Score of Logistic Regression: {precision_logistic}")
recall_logistic = recall_score(y_test, logistic_pred)
print(f"Recall Score of Logistic Regression: {recall_logistic}")
f1_logistic = f1_score(y_test, logistic_pred)
print(f"F1 Score of Logistic Regression: {f1_logistic}")
logistic_cv_scores=cross_val_score(logistic_model,x,y,cv=5)
print("Logistic Regreesion 5-Fold CV Scores:", logistic_cv_scores)
print("Logistic Regression CV Accuracy:", logistic_cv_scores.mean())
cm_decision = confusion_matrix(y_test, decision_pred)
print(f"\nDecision Tree Confusion Matrix:")
print(cm_decision)
precision_decision = precision_score(y_test, decision_pred)
print(f"Precision Score of Decision Tree: {precision_decision}")
recall_decision = recall_score(y_test, decision_pred)
print(f"Recall Score of Decision Tree: {recall_decision}")
f1_decision = f1_score(y_test, decision_pred)
print(f"F1 Score of Decision Tree: {f1_decision}")
decision_cv_scores = cross_val_score(decision_model,x,y,cv=5)
print("Decision Tree 5-Fold CV Scores:", decision_cv_scores)
print("Decision Tree Mean CV Accuracy:", decision_cv_scores.mean())
```

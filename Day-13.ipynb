```python
import pandas as pd
import seaborn as sb
# Load Titanic dataset
titanic = sb.load_dataset("titanic")
# 1. Count missing values
print("Missing values:")
print(titanic.isnull().sum())
# 2. Fill missing age with median
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
# 3. Drop deck because too many values are missing
titanic = titanic.drop("deck", axis=1)
# 4. Fill missing embarked values with mode
titanic["embarked"] = titanic["embarked"].fillna(
    titanic["embarked"].mode()[0])
# 5. Convert sex into numbers
titanic["sex"] = titanic["sex"].map({
    "male": 0,
    "female": 1
})
# 6. Create family size
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1
# 7. Check missing values again
print("\nMissing values after cleaning:")
print(titanic.isnull().sum())
# 8. Look at the cleaned dataset
print("\nCleaned dataset:")
print(titanic.head())
```

```python
#Train a RandomForestClassifier on your training data.
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
x=titanic[["pclass","sex","age","family_size"]]
y=titanic["survived"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
rf=RandomForestClassifier(n_estimators=100,random_state=42)
rf.fit(x_train,y_train)
rf_pred=rf.predict(x_test)
```

```python
#Evaluate it using accuracy, precision, recall, and F1 score, the same way you did in Day 10.
from sklearn.metrics import precision_score, recall_score, f1_score
ps=precision_score(y_test,rf_pred)
print(f"Precision Score: {ps}")
rs=recall_score(y_test,rf_pred)
print(f"Recall Score: {rs}")
f1s=f1_score(y_test,rf_pred)
print(f"F1 Score: {f1s}")
```

```python
#Print the feature importances and identify the two columns that mattered most to the model.
feature_importance=pd.Series( rf.feature_importances_,index=x.columns)
print(feature_importance.sort_values(ascending=False))
```

```python
# Compare the Random Forest's cross validated score against every other model trained this
#week.
from sklearn.model_selection import cross_val_score
rf_cv_score=cross_val_score(rf,x,y,cv=5)
print(f"Random Forest 5 Folds CV Score:{rf_cv_score}")
```

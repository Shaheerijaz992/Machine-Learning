```python

import torch

torch.manual_seed(42)

x = torch.arange(1, 11, dtype=torch.float32)
noise = torch.randn(10) * 0.5
y = 3 * x + 2 + noise

weight = torch.tensor(0.0, requires_grad=True)
bias = torch.tensor(0.0, requires_grad=True)

learning_rate = 0.01

for epoch in range(1000):
    y_pred = weight * x + bias
    loss = torch.mean((y_pred - y) ** 2)
    loss.backward()
    with torch.no_grad():
        weight -= learning_rate * weight.grad
        bias -= learning_rate * bias.grad
    weight.grad.zero_()
    bias.grad.zero_()
    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch: {epoch + 1}, "
            f"Loss: {loss.item():.4f}, "
            f"Weight: {weight.item():.4f}, "
            f"Bias: {bias.item():.4f}")
print("\nFinal Weight:", weight.item())
print("Final Bias:", bias.item())
```

```python
import torch
torch.manual_seed(42)
x = torch.arange(1, 11, dtype=torch.float32)
noise = torch.randn(10) * 0.5
y = 3 * x + 2 + noise
for lr in [0.1, 0.01, 0.001]:
    weight = torch.tensor(0.0, requires_grad=True)
    bias = torch.tensor(0.0, requires_grad=True)
    print(f"\nLearning Rate: {lr}")
    for epoch in range(1000):
        y_pred = weight * x + bias
        loss = torch.mean((y_pred - y) ** 2)
        loss.backward()
        with torch.no_grad():
            weight -= lr * weight.grad
            bias -= lr * bias.grad
        weight.grad.zero_()
        bias.grad.zero_()
        if (epoch + 1) % 100 == 0:
            print(
                f"Epoch: {epoch + 1}, "
                f"Loss: {loss.item():.4f}, "
                f"Weight: {weight.item():.4f}, "
                f"Bias: {bias.item():.4f}"
            )
    print("Final Weight:", weight.item())
    print("Final Bias:", bias.item())
```
